## <font color="#742D2B">1. Definición del movimiento Browniano</font>

**Definición.** Un proceso estocástico $\{W(t):t\in[0,T]\}$ se llama **movimiento Browniano** o **proceso estándar de Wiener** si satisface las siguientes condiciones:

**C₁. Condición inicial**

$$
W(0)=0
$$

**C₂. Incrementos normales**

Para $0\leq s<t\leq T$, el incremento $W(t)-W(s)$ tiene distribución normal con media cero y varianza $t-s$:

$$
W(t)-W(s)\sim\mathcal{N}(0,t-s)
$$

Equivalentemente,

$$
W(t)-W(s)\overset{d}{=}\sqrt{t-s}\,Z,
\qquad Z\sim\mathcal{N}(0,1).
$$

**C₃. Incrementos independientes**

Para $0\leq s\leq t\leq u\leq v\leq T$, los incrementos $W(t)-W(s)$ y $W(v)-W(u)$ son independientes.

## <font color="#742D2B">2. Propiedades del valor esperado y la varianza</font>

Como $W(0)=0$, tomando $s=0$ obtenemos:

$$
W(t)\sim\mathcal{N}(0,t).
$$

Por lo tanto,

$$
\boxed{\mathbb{E}[W(t)]=0}
$$

$$
\boxed{\mathbb{E}[W(t)^2]=t}
$$

$$
\boxed{\operatorname{Var}(W(t))=t}
$$

### <font color="#2A704B">Teorema: valor esperado y varianza</font>

Sean $X,Y$ variables aleatorias con varianzas finitas y $a,b$ constantes.

1. Linealidad del valor esperado:

$$
\mathbb{E}[aX+b]=a\mathbb{E}[X]+b
$$

2. Esperanza de una suma:

$$
\mathbb{E}[X+Y]=\mathbb{E}[X]+\mathbb{E}[Y]
$$

3. Varianza:

$$
\operatorname{Var}(X)=\mathbb{E}[X^2]-\mu^2,
\qquad \mu=\mathbb{E}[X]
$$

4. Varianza de una transformación lineal:

$$
\operatorname{Var}(aX+b)=a^2\operatorname{Var}(X)
$$

## <font color="#742D2B">4. Movimiento Browniano discretizado</font>

Para simular un movimiento Browniano en el intervalo $[0,T]$, dividimos el intervalo en $L$ subintervalos de igual longitud.

**Tamaño del paso:**

$$
\boxed{\Delta t=\frac{T}{L}},\qquad L\in\mathbb{N}.
$$

Los puntos de la partición son:

$$
t_i=i\Delta t,\qquad i=0,1,\ldots,L.
$$

**Condición inicial:**

$$
W_0=0.
$$

**Actualización del proceso:**

$$
W_{i+1}=W_i+\Delta W_i.
$$

Generamos variables normales estándar independientes:

$$
Z_i\sim\mathcal{N}(0,1).
$$

Los incrementos se calculan como:

$$
\boxed{\Delta W_i=\sqrt{\Delta t}\,Z_i}
$$

Por lo tanto, la fórmula de actualización es:

$$
\boxed{W_{i+1}=W_i+\sqrt{\Delta t}\,Z_i}
$$

Cada incremento satisface:

$$
\Delta W_i\sim\mathcal{N}(0,\Delta t).
$$

**Ejercicio:**  Encuentre la probabilidad de que el **MB** con parametro $μ=0.6$ y $σ^2=0.25 $  tome valores entre $1$ y $3$ al tiempo $t=4$

**SOLUCION ANALITICA**

Trabajaremos bajo el supuesto de que se trata de un movimiento Browniano con deriva, definido por
$$
X(t)=\mu t+\sigma W(t),
$$

donde $(W(t))$ es un movimiento Browniano estándar, $(\mu)$ es la deriva y $(\sigma^2)$ es la varianza por unidad de tiempo.
1. Solución analítica
Paso 1. Identificar los datos.
$$
\mu=0.6,\qquad \sigma^2=0.25,\qquad t=4.
$$

Además, queremos calcular:
$$
\boxed{P(1<X(4)<3)}
$$

Paso 2. Determinar la distribución de \(X(t)\).

Para un movimiento Browniano con deriva, se cumple que
$$
X(t)\sim\mathcal{N}(\mu t,\sigma^2t).
$$

Esto significa que la media y la varianza del proceso en el tiempo \(t\) son
$$
\mathbb{E}[X(t)]=\mu t
$$

y
$$
\operatorname{Var}(X(t))=\sigma^2t.
$$

Sustituimos los valores del ejercicio.

Media:
$$
\mathbb{E}[X(4)]=(0.6)(4)=2.4.
$$

Varianza:
$$
\operatorname{Var}(X(4))=(0.25)(4)=1.
$$

Por lo tanto, la distribución en el tiempo \(t=4\) es
$$
\boxed{X(4)\sim\mathcal{N}(2.4,1)}.
$$

La desviación estándar es
$$
\sigma_4=\sqrt{1}=1.
$$

Paso 3. Expresar la probabilidad.

Queremos encontrar
$$
P(1<X(4)<3).
$$

Estandarizamos mediante la transformación
$$
Z=\frac{X(4)-2.4}{1},
$$

donde $(Z\sim\mathcal{N}(0,1))$.

Aplicando la transformación a los dos límites:
$$
P\left(\frac{1-2.4}{1}<Z<
\frac{3-2.4}{1}\right).
$$

Simplificando:
$$
\boxed{P(-1.4<Z<0.6)}.
$$

Paso 4. Utilizar la función de distribución normal estándar.

Denotamos por $(\Phi(z))$ la función de distribución acumulada de una normal estándar. Entonces,
$$
P(-1.4<Z<0.6)=\Phi(0.6)-\Phi(-1.4).
$$

Consultando la tabla de la normal estándar, obtenemos aproximadamente:
$$
\Phi(0.6)=0.7257
$$

$$
\Phi(-1.4)=0.0808.
$$

Por lo tanto,
$$
P(1<X(4)<3)=0.7257-0.0808.
$$

El resultado analítico es
$$
\boxed{P(1<X(4)<3)\approx0.6450}.
$$

En porcentaje, la probabilidad es aproximadamente 64.50 %.
